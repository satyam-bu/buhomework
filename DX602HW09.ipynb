**Problem 2 (review)**

**Concept:** *reading CSV file, create an np.array of numbers*

**Task:**

Write a function `read_csv_to_dict(filename)` that reads a CSV file and stores the data in a dictionary.

Your function should:
1. Open the CSV file.
2. Read the first row to obtain column headers.
3. Store each column of data as a NumPy array.
4. Use the column names as keys in the dictionary.
5. Return the dictionary.

Each key should correspond to one column in the CSV file.

Example structure:

```
{
   "open": np.array([...]),
   "close": np.array([...]),
   "volume": np.array([...])
}
```


````python
#Your Solution

def read_csv_to_dict(filename):
    """Read a CSV file and return a dict mapping each column name to a NumPy array."""
    # Open file
    with open(filename, "r") as file:
        # Read the first row as headers
        headers = file.readline().strip().split(",")

        # Create an empty list for each column
        columns = [[] for _ in headers]

        # Loop through each remaining row in the file
        for line in file:
            line = line.strip()
            
            if line == "":
                continue
            # Split the row into values
            values = line.split(",")
            # Add each value to its column list
            for i in range(len(headers)):
                columns[i].append(values[i])

    # Build the dictionary: column name -> NumPy array
    data = {}
    for i in range(len(headers)):
        data[headers[i]] = np.array(columns[i])

    return data


````

**Task:**

Create a scatter plot that displays the points of a **unit circle**.

Steps:

1. Generate `x` values between **0 and 1**.
2. Compute the corresponding `y` values using the circle equation:

$$y = sqrt(1 − x²)$$

3. Reflect the points across all four quadrants so that the full circle is drawn.
4. Plot the points using `plt.scatter`.

Set the figure size to **(5,5)** so the circle appears correctly proportioned.

````python
#Your Solution

# Generate x values between 0 and 1
x = np.arange(0, 1, 0.01)
# Compute y from the circle equation 
y = np.sqrt(1 - x**2)

# Reflect the points across all four quadrants

full_x = np.concatenate((x, -x, -x, x))
full_y = np.concatenate((y, y, -y, -y))

# Plot the points with a square figure 
plt.figure(figsize=(5, 5))
plt.scatter(full_x, full_y, s=10)
plt.title("Unit Circle")
plt.xlabel("x")
plt.ylabel("y")
````

**Task:**

Write a function:

`create_line_graph(filename, column_name)`

Your function should:
1. Read the CSV file using `read_csv_to_dict`.

3. Convert the column values to floats.
4. Plot the values as a **line graph**.

````python
#Your Solution

def create_line_graph(filename, column_name):
    """Read a CSV file and plot one column as a line graph against day number."""
    # Read the CSV file into a dictionary of NumPy arrays
    data = read_csv_to_dict(filename)

    # Verify the specified column exists
    if column_name not in data:
        print(f"Column '{column_name}' not found in {filename}.")
        return

    # Convert the column values from strings to floats
    values = data[column_name].astype(float)

    # Day numbers start at 1 and match the number of values
    days = np.arange(1, len(values) + 1)

    # Plot the values as a line graph
    plt.figure(figsize=(10, 5))
    plt.plot(days, values)
    plt.title(f"{column_name} by Day")
    plt.xlabel("Day Number")
    plt.ylabel(column_name)
````

**Problem 12**

**Concepts:** *reading CSV file, pie chart*

**Task:**

Write a function:

`create_pie_chart(filename, column_name)`

Your function should:
1. Read the CSV file using `read_csv_to_dict`.
2. Convert the selected column to numeric values.
3. Compute the **mean value** of the column.
4. Determine:
* the percentage of values **greater than the mean**
* the percentage **less than or equal to the mean**

Create a pie chart displaying these two proportions.

Include:
* labels
* percentages
* chart title

````python
#Your Solution

def create_pie_chart(filename, column_name):
    """Read a CSV file and plot a pie chart of values above vs. at/below the column mean."""
    # Read the CSV file into a dictionary of NumPy arrays
    data = read_csv_to_dict(filename)

    # Verify the specified column exists
    if column_name not in data:
        print(f"Column '{column_name}' not found in {filename}.")
        return

    # Convert the column values from strings to floats
    values = data[column_name].astype(float)

    # empty column check
    if len(values) == 0:
        print(f"Column '{column_name}' has no values.")
        return

    # Compute mean
    mean_value = np.mean(values)

    # Count values greater than the mean
    above = int(np.sum(values > mean_value))
    at_or_below = len(values) - above

    # Plot the pie chart
    plt.figure(figsize=(6, 6))
    plt.pie(
        [above, at_or_below],
        labels=["Greater than mean", "Less than or equal to mean"],
        autopct="%1.1f%%",
        colors=["tab:blue", "tab:orange"],
    )
    plt.title(f"{column_name}: Values Relative to Mean ({mean_value:.2f})")
````

**Problem 13**

**Concepts:** *reading CSV file, scatter plot*

**Task:**

Write a function:

`create_scatter_plot(filename, x_column, y_column)`

Your function should:
1. Read the CSV file using `read_csv_to_dict`.
2. Extract the specified columns.
3. Convert the values to floats.
4. Create a scatter plot showing the relationship between the two variables.

Include:
* x-axis label
* y-axis label
* title

````python
#Your Solution

def create_scatter_plot(filename, x_column, y_column):
    """Read a CSV file and plot two columns against each other as a scatter plot."""
    # Read the CSV file into a dictionary of NumPy arrays
    data = read_csv_to_dict(filename)

    # Verify both columns exist
    if x_column not in data or y_column not in data:
        print(f"Column '{x_column}' or '{y_column}' not found in {filename}.")
        return

    # Convert the column values from strings to floats
    x = data[x_column].astype(float)
    y = data[y_column].astype(float)

    # Plot the two variables as a scatter plot
    plt.figure(figsize=(8, 6))
    plt.scatter(x, y, s=15)
    plt.title(f"{y_column} vs. {x_column}")
    plt.xlabel(x_column)
    plt.ylabel(y_column)
````
